# Analisis Dataset "E-Commerce Order & Supply Chain"

Dataset ini berisi data transaksi e-commerce yang mencakup informasi mengenai pesanan, produk, pelanggan, pembayaran, seller, serta proses pengiriman. Analisis ini bertujuan untuk memahami performa bisnis dan operasional e-commerce, khususnya dalam melihat pola penjualan, performa produk dan seller, metode pembayaran, serta efektivitas proses pengiriman kepada pelanggan.

Dataset terdiri dari beberapa tabel yang saling berhubungan melalui `order_id`, `customer_id`, dan `product_id`.

Beberapa dataset dan kolom penting yang digunakan dalam analisis ini meliputi:

- **Orders** : Informasi utama mengenai pesanan dan proses pengiriman.
  - `order_id` : ID unik setiap pesanan.
  - `customer_id` : ID pelanggan yang melakukan pesanan.
  - `order_status` : Status pesanan.
  - `order_purchase_timestamp` : Waktu ketika pesanan dibuat.
  - `order_delivered_customer_date` : Waktu ketika pesanan diterima pelanggan.
  - `order_estimated_delivery_date` : Estimasi waktu pengiriman.

- **Order Items** : Informasi produk dan seller pada setiap transaksi.
  - `order_id` : ID pesanan.
  - `product_id` : ID produk.
  - `seller_id` : ID seller.
  - `price` : Harga produk.
  - `shipping_charges` : Biaya pengiriman.

- **Products** : Informasi mengenai produk yang dijual.
  - `product_id` : ID unik produk.
  - `product_category_name` : Kategori produk.
  - Informasi dimensi dan berat produk.

- **Customers** : Informasi lokasi pelanggan.
  - `customer_id` : ID pelanggan.
  - `customer_city` : Kota pelanggan.
  - `customer_state` : State atau wilayah pelanggan.

- **Payments** : Informasi mengenai metode pembayaran transaksi.
  - `order_id` : ID pesanan.
  - `payment_type` : Metode pembayaran.
  - `payment_installments` : Jumlah cicilan.
  - `payment_value` : Nilai pembayaran.

Dalam analisis ini, kita akan mengeksplorasi hubungan antar tabel untuk mengidentifikasi pola penjualan, performa kategori produk dan seller, tren transaksi, performa pengiriman, serta perbedaan performa operasional berdasarkan wilayah pelanggan.

Proses analisis akan menggunakan **Python melalui Google Colab** untuk data understanding, data cleaning, feature engineering, exploratory data analysis, dan visualisasi. Selanjutnya, **Google BigQuery dan SQL** akan digunakan untuk melakukan pengolahan data relasional, agregasi, serta analisis bisnis yang lebih mendalam.

## Data Preparation

Pada tahap ini dilakukan proses persiapan dan pembersihan data agar dataset siap digunakan untuk analisis lebih lanjut. Proses ini mencakup penyesuaian tipe data, penanganan duplikasi dan missing values, validasi data, serta pembuatan variabel baru yang dibutuhkan dalam analisis.

```python
# Create Copy Dataset
orders_clean = orders.copy()
order_items_clean = order_items.copy()
products_clean = products.copy()
customers_clean = customers.copy()
payments_clean = payments.copy()
```

```python
# Changing Data Type of Date Column
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_timestamp",
    "order_estimated_delivery_date"
]

for column in date_columns:
    orders_clean[column] = pd.to_datetime(
        orders_clean[column],
        errors="coerce"
    )
```

```python
# Date Column Data Type Validation
orders_clean[date_columns].dtypes
```

```python
# Handling Duplicate in Product
product_check = (
    products_clean
    .groupby("product_id")
    .nunique(dropna=False)
)

conflicting_products = product_check[
    (product_check > 1).any(axis=1)
]

print("Jumlah product_id dengan informasi yang berbeda:",
      len(conflicting_products))
```

```python
# Drop Duplicate in Product
products_clean = products_clean.drop_duplicates(
    subset="product_id"
).reset_index(drop=True)
```

```python
# Check Validation
print("Jumlah baris setelah cleaning:", len(products_clean))
print("Jumlah product_id unique:", products_clean["product_id"].nunique())
print("Jumlah duplicate product_id:",
      products_clean["product_id"].duplicated().sum())
```

Hasil menunjukkan bahwa tidak terdapat `product_id` yang memiliki informasi produk berbeda. Dengan demikian, pengulangan pada dataset `Products` merupakan data duplikat yang konsisten.

Data duplikat kemudian dihapus berdasarkan `product_id`, sehingga jumlah data berkurang dari 89.316 menjadi 27.451 baris dan setiap `product_id` hanya memiliki satu record unik.

Proses ini penting untuk mencegah terjadinya penggandaan nilai ketika dataset `Products` digabungkan dengan dataset transaksi pada tahap analisis.

```python
# Handling Missing Values
for name, df in {
    "Orders": orders_clean,
    "Order Items": order_items_clean,
    "Products": products_clean,
    "Customers": customers_clean,
    "Payments": payments_clean
}.items():
    print(f"\n{name}")
    print(df.isnull().sum())
    print(f"{'='*50}")
```

```python
# Check Missing Values in order_delivered_timestamp by Categorical order_status
pd.crosstab(
    orders_clean["order_status"],
    orders_clean["order_delivered_timestamp"].isna()
)
```

Sebagian besar missing values pada `order_delivered_timestamp` berasal dari pesanan yang belum mencapai status delivered atau telah dibatalkan. Oleh karena itu, missing values tersebut dianggap sesuai dengan kondisi transaksi dan tetap dipertahankan sebagai `NaT`.

Namun, terdapat 6 pesanan dengan status `delivered` yang tidak memiliki `order_delivered_timestamp`. Data tersebut akan tetap dipertahankan, tetapi tidak digunakan dalam analisis durasi pengiriman karena tidak memiliki informasi waktu pengiriman yang lengkap.

Terdapat 9 pesanan berstatus `delivered` yang tidak memiliki informasi `order_approved_at`. Missing values pada `product_category_name` diisi dengan kategori `Unknown`. Hal tersebut digunakan agar produk tetap dapat dipertahankan dalam dataset tanpa mengasumsikan kategori produk yang sebenarnya.

```python
# Check Missing Values in Product by Dimension
dimension_columns = [
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

products_clean[
    products_clean[dimension_columns].isnull().any(axis=1)
]
```

Terdapat 2 produk yang tidak memiliki informasi berat dan dimensi. Karena produk tersebut masih memiliki informasi valid yang dapat digunakan untuk analisis transaksi, kategori, dan penjualan, record tetap dipertahankan dalam dataset utama.

Produk dengan missing values hanya akan dikecualikan pada analisis yang secara khusus membutuhkan informasi berat atau dimensi produk.

```python
# Validation Numeric Values
numeric_checks = {
    "Order Items": (
        order_items_clean,
        ["price", "shipping_charges"]
    ),
    "Products": (
        products_clean,
        [
            "product_weight_g",
            "product_length_cm",
            "product_height_cm",
            "product_width_cm"
        ]
    ),
    "Payments": (
        payments_clean,
        ["payment_installments", "payment_value"]
    )
}

for name, (df, columns) in numeric_checks.items():
    print(f"\n{name}")

    for column in columns:
        print(
            f"{column}: "
            f"min={df[column].min()}, "
            f"max={df[column].max()}, "
            f"zero={(df[column] == 0).sum()}, "
            f"negative={(df[column] < 0).sum()}"
        )
```

Karena nilai tersebut tidak merepresentasikan berat produk yang realistis, nilai 0 diganti menggunakan median product_weight_g dari produk dalam kategori toys. Pendekatan ini digunakan agar nilai pengganti lebih merepresentasikan karakteristik produk sejenis tanpa menghapus record produk dari dataset.

## Feature Engineering  

Pada tahap ini dibuat beberapa variabel baru berdasarkan data yang tersedia untuk mendukung analisis transaksi, waktu, dan performa pengiriman. Variabel baru dibuat tanpa mengubah informasi asli pada dataset.

```python
# Create Time Feature in Orders
orders_clean["purchase_year"] = (
    orders_clean["order_purchase_timestamp"].dt.year
)

orders_clean["purchase_month"] = (
    orders_clean["order_purchase_timestamp"]
    .dt.to_period("M")
    .astype(str)
)

orders_clean["purchase_day"] = (
    orders_clean["order_purchase_timestamp"]
    .dt.day_name()
)
```

```python
# Check Results
orders_clean[
    [
        "order_purchase_timestamp",
        "purchase_year",
        "purchase_month",
        "purchase_day"
    ]
].head()
```

Variabel `purchase_year`, `purchase_month`, dan `purchase_day` dibuat dari `order_purchase_timestamp` untuk memudahkan analisis tren transaksi berdasarkan tahun, bulan, dan hari.

```python
# Create Delivery Time
orders_clean["delivery_days"] = (
    orders_clean["order_delivered_timestamp"]
    - orders_clean["order_purchase_timestamp"]
).dt.total_seconds() / 86400
```

```python
# Check Results
orders_clean[
    [
        "order_purchase_timestamp",
        "order_delivered_timestamp",
        "delivery_days"
    ]
].head()
```

Variabel `delivery_days` menunjukkan durasi aktual pengiriman sejak order dibuat hingga diterima oleh customer. Order yang belum memiliki waktu pengiriman tetap memiliki nilai kosong dan tidak digunakan dalam analisis durasi pengiriman.

```python
# Create Delivery Status
orders_clean["delivery_status"] = np.select(
    [
        orders_clean["order_delivered_timestamp"].isna(),
        orders_clean["order_delivered_timestamp"]
        <= orders_clean["order_estimated_delivery_date"]
    ],
    [
        "Not Delivered",
        "On Time"
    ],
    default="Late"
)
```

```python
# Check Results
orders_clean["delivery_status"].value_counts()
```

Variabel `delivery_status` dibuat untuk mengelompokkan performa pengiriman menjadi `On Time`, `Late`, dan `Not Delivered`.

Pesanan dikategorikan sebagai `On Time` apabila waktu pengiriman aktual tidak melewati estimasi pengiriman, `Late` apabila melewati estimasi, dan `Not Delivered` apabila belum memiliki waktu pengiriman aktual.

```python
# Create Order Values
order_items_clean["order_value"] = (
    order_items_clean["price"]
    + order_items_clean["shipping_charges"]
)
```

```python
# Check Results
order_items_clean[
    [
        "price",
        "shipping_charges",
        "order_value"
    ]
].head()
```

Variabel `order_value` dihitung dari penjumlahan harga produk (`price`) dan biaya pengiriman (`shipping_charges`). Variabel ini akan digunakan untuk melihat hasil transaksi setiap order.

Sementara itu, `price` tetap digunakan sebagai nilai penjualan produk agar biaya pengiriman dapat dianalisis secara terpisah.

```python
%%bigquery --project {project_id} --verbose

# Monthly Sales Performance
WITH monthly_order_data AS (
  SELECT
    o.order_id,
    FORMAT_DATE('%Y-%m', o.order_purchase_timestamp) AS month_key,
    oi.price,
    oi.order_value,
    o.order_status
  FROM ecommerce.orders AS o
  LEFT JOIN ecommerce.order_items AS oi
    ON o.order_id = oi.order_id
)
SELECT
  FORMAT_DATE('%Y - %B', PARSE_DATE('%Y-%m', month_key)) AS purchase_month,
  COUNT(DISTINCT order_id) AS total_orders,
  ROUND(
    SUM(price),
    2
  ) AS total_sales,
  ROUND(
    AVG(order_value),
    2
  ) AS average_order_value
FROM monthly_order_data
WHERE order_status = 'delivered'
GROUP BY
  month_key
ORDER BY
  PARSE_DATE('%Y-%m', month_key);
```

```python
%%bigquery --project {project_id} --verbose

# Monthly Sales Growth
WITH monthly_sales AS (

SELECT
  FORMAT_DATE('%Y-%m', o.order_purchase_timestamp) AS month_key,

  SUM(oi.price) AS total_sales

FROM ecommerce.orders AS o

LEFT JOIN ecommerce.order_items AS oi
ON o.order_id = oi.order_id

WHERE o.order_status = 'delivered'

GROUP BY
  month_key

)

SELECT

  FORMAT_DATE('%Y - %B', PARSE_DATE('%Y-%m', month_key)) AS purchase_month,

  ROUND(total_sales,2) AS total_sales,

  ROUND(
    LAG(total_sales)
    OVER(
      ORDER BY PARSE_DATE('%Y-%m', month_key)
    ),
    2
  ) AS previous_month_sales,


  ROUND(
    SAFE_DIVIDE(
      total_sales -
      LAG(total_sales)
      OVER(
        ORDER BY PARSE_DATE('%Y-%m', month_key)
      ),

      LAG(total_sales)
      OVER(
        ORDER BY PARSE_DATE('%Y-%m', month_key)
      )

    ) * 100,
    2
  ) AS growth_percentage


FROM monthly_sales

ORDER BY PARSE_DATE('%Y-%m', month_key);
```

## Advanced Analytics  

1. How are customers distributed based on spending level?
2. Do a small number of sellers contribute a large portion of revenue?
3. Which products generate the highest sales performance?

```python
%%bigquery --project {project_id} --verbose

# Customer Spending Segmention
WITH customer_spending AS (

SELECT

  o.customer_id,

  SUM(p.payment_value) AS total_spending

FROM ecommerce.orders AS o

JOIN ecommerce.payments AS p

ON o.order_id = p.order_id

WHERE
  o.order_status = 'delivered'

GROUP BY
  o.customer_id

)

SELECT

CASE

  WHEN total_spending < 100 THEN 'Low Spender'

  WHEN total_spending BETWEEN 100 AND 500 THEN 'Medium Spender'

  ELSE 'High Spender'

END AS spending_segment,

COUNT(customer_id) AS total_customers

FROM customer_spending

GROUP BY
  spending_segment

ORDER BY
  total_customers DESC;
```

```python
%%bigquery --project {project_id} --verbose

# Pareto Cumulative Analysis
WITH seller_sales AS (

SELECT

  seller_id,
  SUM(price) AS total_sales

FROM ecommerce.order_items

GROUP BY
  seller_id

),

seller_rank AS (

SELECT

  seller_id,
  total_sales,

  SUM(total_sales)
  OVER(
    ORDER BY total_sales DESC
  ) AS cumulative_sales,

  SUM(total_sales)
  OVER() AS overall_sales

FROM seller_sales

)

SELECT

  seller_id,
  ROUND(total_sales,2) AS total_sales,
  ROUND(
    cumulative_sales /
    overall_sales * 100,
    2
  ) AS cumulative_percentage

FROM seller_rank
ORDER BY
  total_sales DESC;
```

```python
%%bigquery --project {project_id} --verbose

# Product Performance Ranking
SELECT

  oi.product_id,
  p.product_category_name,
  COUNT(oi.order_id) AS total_orders,

  ROUND(
    SUM(oi.price),
    2
  ) AS total_sales,

  RANK()
  OVER(
    ORDER BY SUM(oi.price) DESC
  ) AS sales_rank

FROM ecommerce.order_items AS oi

LEFT JOIN ecommerce.products AS p

ON oi.product_id = p.product_id

GROUP BY
  oi.product_id,p. product_category_name

ORDER BY
  sales_rank

LIMIT 20;
```
